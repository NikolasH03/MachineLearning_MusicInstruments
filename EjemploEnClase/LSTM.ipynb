```python
from tensorflow.keras.preprocessing import sequence

LONG_MAX_SEC = 100
X_train_pad = sequence.pad_sequences(x_train, maxlen=LONG_MAX_SEC, padding='post', truncating='post')
x_test_pad = sequence. pad_sequences(x_train, maxlen=LONG_MAX_SEC, padding='post', truncating='post')
print(X_train_pad[159,:])
```

**Embedding** tranforma a numero conjuntos cada numero representa una dimension. (vector)
No es una conversion aleatoria sino info de contenido

```python
# Enbeddings de 128 dimensiones (cada palabra esta representada por 128)
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding
import tensorflow as tf

SEED = 32
tf.random.set_seed(SEED) #para que repita lo que corre
np.random.seed(SEED)

#parámetros embedding
DIM_IN = TAM_VOCAB
DIM_OUTPUT = 128
modelo = Sequential()
modelo.add(Embedding(input_dim=DIM_IN, output_dim=DIM_OUTPUT, input_length=LONG_MAX_SEC))
```

```python
#LSTM y capa oculta
#LSTM y capa salida
from tensorflow.keras.layers import LSTM, Dense
TAM_OCULTO = 64 #celda de memoria c_t y el estado oculto h_t-1
#TAM_OCULTO es muy importante porque es cuanto va a recordar y olvidar
modelo.add(LSTM(TAM_OCULTO))

modelo.add(Dense(1, activation='sigmoid'))
```

```python
modelo.build((None, LONG_MAX_SEC))
modelo.summary()
```

100 longitud maxima, 128 tamaño del vector de cada palabra
64 tamaño oculto de la memoria oculta reapartido en varias mini redes neuronales


```python
from tensorflow.keras.optimizers import RMSprop
optimizador = RMSprop(learning_rate = 0.00005) #esta tasa de aprendizaje es muy baja y tiende al sobreajuste
modelo.compile(loss='binary_crossentropy', optimizer=optimizador, metrics=['accuracy'])

tam_batch = 50
epocas = 5
historia = modelo.fit(X_train_pad, y_train, batch_size=tam_batch, epochs=epocas, validation_data=(x_test_pad, y_test))
#y_entre etiquetas con las que compara el error
```
