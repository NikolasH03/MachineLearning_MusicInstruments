```python
#Definir una politica (estrategia del agente pa llegar al objetivo final)
#aleatoria: la prob de llegar a la acción a partiendo del estado s: igual para las 4 acciones
import numpy as np
politica_al= np.ones([entorno.nS,entorno.nA])/entorno.nA
print(politica_al.shape)
print(politica_al)
#columnas son acciones y filas estados
```

evaluacion de la politica:

entrada: entorno, la politica (nSxnA), gamma, theta (crietrio de convergencia

salida: l funcion v(1xnS)

```python
#constantes
GAMMA = 0.99
THETA = 1e-10
```

```python
#algoritmo de la evaluación de la politica
def evaluacion_politica(entorno, politica, gamma=GAMMA, theta=THETA):
  V = np.zeros(entorno.nS)

  while True:
    delta = 0
    for s in range(entorno.nS):
      Vs= 0
      for a, pr in enumerate(politica[s]):
        for prob, state_, reward, done in entorno.P[s][a]:
          Vs += pr * prob * (reward + gamma * V[state_])
      delta = max(delta, np.abs(V[s]- Vs))
      V[s] = Vs

    if delta < theta: #si converge deteiene las iteraciones
      break

  return V
```

```python
V_al = evaluacion_politica(entorno, politica_al)
print('Evaluacion de la politica aleatoria')
print(V_al.reshape(4,4))
```

Mejorar la politica:

entrada: entorno, politica, la funcionnde eestado de valor, gamma

Salida: q

```python
def calcular_funcion_q(entorno, V, s, gamma=GAMMA):
  q = np.zeros(entorno.nA)
  for a in range(entorno.nA):
    for prob, state_, reward, done in entorno.P[s][a]:
      q[a] += prob*(reward + GAMMA * V[state_])
  return q
```

```python
Q_al = np.zeros([entorno.nS, entorno.nA])

for s in range(entorno.nS):
  Q_al[s] = calcular_funcion_q(entorno, V_al, s)

print('Funcion accion-valor politica aleatoria')
print(Q_al)
```

```python
def mejorar_politica(entorno, V, gamma=GAMMA):
  #se inicia con una politica arbitraria
  politica = np.zeros([entorno.nS, entorno.nA])

  #Y se mejora
  for s in range(entorno.nS):
    q = calcular_funcion_q(entorno, V, s, gamma)
    politica[s][np.argmax(q)] = 1
  return politica
```

```python
politica_mejorada = mejorar_politica(entorno, V_al)
print('Politica Alearoria')
print(politica_al)
print('Evaluacion politica')
print(V_al)
print('Politica Mejorada')
print(politica_mejorada)
```

Iteracion de la politica (obtener la politica optima)

entrada: entorno, gamma, theta

salida: politica optima (nSxnA)

```python
def iterar_politica(entorno, gamma = GAMMA, theta=THETA):
  politica = np.ones([entorno.nS, entorno.nA])/entorno.nA
  while True:
    V = evaluacion_politica(entorno, politica)
    politica_nueva = mejorar_politica(entorno, V)
    if (politica_nueva == politica).all():
      break
    politica = np.copy(politica_nueva)
  return politica_nueva, V
```

```python
politica_ite, V_ite = iterar_politica(entorno)
print('Politica Optima iteracion de la politica')
print(politica_ite)
print('Politica original')
print(politica_al)
print('Funcion estado valor - iteracion de la politica')
print(V_ite.reshape(4,4))
print('Funcion estado valor original')
print(V_al.reshape(4,4))
```

iteracion de valores (politica optima)

entrada: entorno, gamma, theta

salida: politica optima, funcion de estado valor (V)

```python
def iteracion_valores(entorno, gamma=GAMMA, theta=THETA):
  V = np.zeros(entorno.nS)
  #obtener funcion estado - valor optima
  while True:
    delta = 0
    for s in range(entorno.nS):
      v = V[s] #almacenamos el valor anterior
      V[s] = max(calcular_funcion_q(entorno, V, s))
      delta = max(delta,abs(V[s]-v))
    if delta < theta:
      break

  politica = mejorar_politica(entorno, V)
  return politica, V
```

```python
politica_iterVal, V_iterVal = iteracion_valores(entorno)
print('Politica optima iteracion de valores')
print(politica_iterVal)
```

```python
#interactuar con el entorno

s = entorno.reset()
for i in range(100):
  print(f'Time step: {i+1}')
  a = np.argmax(politica_iterVal[s])
  s_,r,done, info = entorno.step(a)
  s = s_
  entorno.render()
  if done:
    if s==15:
      print('El agente llego a la meta')
    else:
      print('El agente se encuentra en estado terminal')
    break
  entorno.close()
```

```python
#evaluar la politica por episodios
def evaluar_politica_episodios(entorno, politica, episodios=100):
  exitos = 0
  for ep in range(episodios):
    s = entorno.reset()
    done = False

    while not done:
      a = np.argmax(politica[s]) #elige la accion que parezca mejor en el momento actual, segun politica
      s,r, done, _ = entorno.step(a)

      if s == 15:
        exitos += 1

  tasa_exito = exitos / episodios
  print(f'Tasa de exito en {episodios} episodios es: {tasa_exito:.2%}')
  return tasa_exito

```

```python
evaluar_politica_episodios(entorno, politica_iterVal, episodios = 1000)
```
